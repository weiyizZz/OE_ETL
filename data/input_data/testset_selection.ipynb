```python
import csv
import random
from collections import defaultdict

# --- 1. Read the table ---
with open('notegroups.csv', newline='') as f:
    rows = list(csv.DictReader(f))

# --- 2. Split off the training set (first 23) and the remaining pool ---
TRAIN_CUTOFF = 23
remaining_rows = [r for r in rows if int(r['notegroupID']) > TRAIN_CUTOFF]

# --- 3. Group remaining notegroups by project ---
proj = defaultdict(list)
for r in remaining_rows:
    proj[r['project_name']].append(int(r['notegroupID']))

projects = list(proj.items())  # [(project_name, [notegroupID, ...]), ...]

# --- 4. Shuffle and accumulate whole projects toward the target ---
SEED = 1          # change this to get a different random draw
TARGET_MIN, TARGET_MAX = 18, 24

random.seed(SEED)
random.shuffle(projects)

selected = []
total = 0
for name, ids in projects:
    if total + len(ids) > TARGET_MAX:
        continue  # adding this project would overshoot too much; skip and try the next
    selected.append((name, ids))
    total += len(ids)
    if total >= TARGET_MIN:
        break  # we've reached an acceptable total, stop here

# --- 5. Report the result ---
print(f"Selected {total} notegroups from {len(selected)} projects (seed={SEED}):\n")
for name, ids in selected:
    print(f"  {name}: {len(ids)} notegroups -> {ids}")
```
