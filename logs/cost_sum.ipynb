```python
import pandas as pd
LOG_PATH = "token_usage_transformer.jsonl"
df = pd.read_json(LOG_PATH, lines=True)
```

```python
total_cost = df["cost_usd"].sum()
print(f"  Total cost:    ${total_cost:.6f}")
```

```python
from datetime import datetime, timezone
df['timestamp'] = pd.to_datetime(df['timestamp'], utc=True)

today = datetime.now(timezone.utc).date()
today_df = df[df['timestamp'].dt.date == today]

print(f"Date: {today}  |  API calls: {len(today_df)}")
print(f"Total cost today: ${today_df['cost_usd'].sum():.6f}\n")
```

```python
import pandas as pd
from datetime import datetime, timezone

EVAL_LOG_PATH = "token_usage_evaluator.jsonl"
df_eval = pd.read_json(EVAL_LOG_PATH, lines=True)
df_eval['timestamp'] = pd.to_datetime(df_eval['timestamp'], utc=True)

since = datetime.fromisoformat("2026-06-16T10:49:29.821261+00:00")
filtered = df_eval[df_eval['timestamp'] >= since]

print(f"API calls since {since}: {len(filtered)}")
print(f"Total cost: ${filtered['cost_usd'].sum():.6f}")
```
